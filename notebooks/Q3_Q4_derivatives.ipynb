## Theory
- Forward difference: $(\partial_x X)_{i,j} = X_{i,j+1} - X_{i,j}$
- Forward difference: $(\partial_y X)_{i,j} = X_{i+1,j} - X_{i,j}$
- Gradient magnitude: $G_{i,j} = \sqrt{(\partial_x X)^2_{i,j} + (\partial_y X)^2_{i,j}}$

## Build Derivative Matrices and Compute Gradients

```python
def compute_and_visualize_gradient(X, name="X"):
    """
    Compute derivatives and gradient magnitude for a 2D field.
    """
    M, N = X.shape
    
    # Build derivative matrices
    Dx = build_derivative_matrix_2d(M, N, 'x')
    Dy = build_derivative_matrix_2d(M, N, 'y')
    
    # Convert to column-stacked vector
    x = grid_to_vector(X)
    
    # Compute derivatives
    Dx_x = Dx @ x
    Dy_x = Dy @ x
    
    # Compute gradient magnitude
    G = compute_gradient_magnitude(Dx_x, Dy_x, M, N)
    
    # Visualize
    visualize_gradients(X, Dx_x, Dy_x, G, 
                       title=f"Gradient Analysis: {name}",
                       save_path=f'../results/figures/{name}_gradients.png')
    
    return Dx_x, Dy_x, G
```

### Process X1

```python
Dx_x1, Dy_x1, G1 = compute_and_visualize_gradient(X1, "X1")
```
